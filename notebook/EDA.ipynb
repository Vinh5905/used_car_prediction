## **Xử lí format price**

```python
import re
```

```python
def parse_vnd_amount(text: str) -> int:
    text = text.strip().lower()
    match = re.match(r"([\d.,]+)\s*(.*)", text)
    if not match:
        raise ValueError(f"Không hiểu được giá trị: {text}")

    number_part, unit_part = match.groups()
    number_part = number_part.replace(',', '.')
    
    try:
        value = float(number_part)
    except ValueError:
        raise ValueError(f"Sai định dạng số: {number_part}")

    if "tỷ" in unit_part or "ty" in unit_part:
        value *= 1_000_000_000
    elif "triệu" in unit_part or "trieu" in unit_part:
        value *= 1_000_000
    elif "nghìn" in unit_part or "ngàn" in unit_part:
        value *= 1_000
    
    return int(value)
```

```python
df['price'] = df['price'].apply(parse_vnd_amount)
```

## **Xử lí mileage**

```python
def parse_mileage(text: str) -> int:
    """
    Chuyển '20 Km' hoặc '48,000 Km' → số km dạng int.
    Trả về None nếu không parse được.
    """
    if not isinstance(text, str):
        return None
    
    # Bỏ chữ 'km', 'Km', 'KM' và khoảng trắng
    text = text.strip().lower().replace('km', '').strip()
    
    # Xóa dấu phẩy hoặc dấu chấm phân cách hàng nghìn
    text = re.sub(r'[.,]', '', text)
    
    # Nếu chuỗi rỗng → bỏ qua
    if not text:
        return None

    try:
        return int(text)
    except ValueError:
        return None
```

```python
df['mileage'] = df['mileage'].apply(parse_mileage)
# Xóa những dòng có mileage = 0
df = df[df['mileage'] != 0]
```

## **Xử lí body_style**

```python
import unicodedata
```

```python
def normalize_vietnamese(text):
    if not isinstance(text, str):
        return text
    # Chuẩn hóa Unicode về dạng NFC
    return unicodedata.normalize('NFC', text.strip())

def normalize_body_style(style):
    if not isinstance(style, str):
        return None

    style = unicodedata.normalize('NFC', style.strip().lower())

    mapping = {
        'suv': 'SUV',
        'sedan': 'Sedan',
        'crossover': 'Crossover',
        'hatchback': 'Hatchback',
        'bán tải / pickup': 'Pickup',
        'van/minivan': 'Minivan',
        'coupe': 'Coupe',
        'convertible/cabriolet': 'Cabriolet',
        'truck': 'Truck',
        'wagon': 'Wagon'
    }

    return mapping.get(style, style.capitalize())
```

```python
df['body_style'] = df['body_style'].apply(normalize_body_style)
```

## **Xử lí engine**

```python
def extract_engine_type(text):
    if not isinstance(text, str):
        return None
    # Lấy từ đầu tiên (loại nhiên liệu), bỏ phần dung tích
    # Regex: match từ đầu chuỗi đến trước số
    match = re.match(r"([A-Za-zÀ-ỹ]+)", text.strip())
    if match:
        return match.group(1).capitalize()
    return None
```

```python
df['engine'] = df['engine'].apply(extract_engine_type)
```

## **Xử lí exterior_color + interior_color**

```python
def normalize_color(value):
    if not isinstance(value, str):
        return None
    value = value.strip().lower()
    
    if value in ['màu khác', '-']:
        return 'Khác'
    
    # Viết hoa chữ cái đầu cho các màu bình thường
    return value.capitalize()
```

```python
df['exterior_color'] = df['exterior_color'].apply(normalize_color)
df['interior_color'] = df['interior_color'].apply(normalize_color)
```

## **Xử lí drivetrain**

```python
def extract_drivetrain_code(value):
    if not isinstance(value, str):
        return None
    match = re.match(r'([A-Za-z0-9]+)', value.strip())
    if match:
        return match.group(1).upper()
    return None

# Bảng mapping ý nghĩa (phục vụ cho việc hiểu value)
drivetrain_mapping = {
    'FWD': 'Dẫn động cầu trước',
    'RWD': 'Dẫn động cầu sau',
    'AWD': '4 bánh toàn thời gian',
    '4WD': 'Dẫn động 4 bánh'
}
```

```python
df['drivetrain'] = df['drivetrain'].apply(extract_drivetrain_code)
```
